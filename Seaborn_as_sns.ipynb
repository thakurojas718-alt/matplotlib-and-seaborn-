```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

# relational plot

```python
# relational plot

tips=sns.load_dataset('tips')
tips
```

```python
# scatterplot-> axis level function
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size')
```

```python
# rel plot-> figure level is used to visiualize relationship betwen two or more variable
sns.relplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size',kind='scatter')
```

```python
# line plot
sns.relplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size',kind='line')
```

```python
sns.lineplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size')
```

```python
gap=px.data.gapminder()
temp_df=gap[gap['country']=='India']
temp_df
```

```python
# axis level function
sns.lineplot(data=temp_df,x='year',y='lifeExp',hue='continent')
```

```python
sns.relplot(data=temp_df,x='year',y='lifeExp',hue='continent',kind='line')
```

```python
df=gap[gap['country'].isin(['India','Brazil','Germany'])]
df
```

```python
sns.relplot(kind='line',data=df,x='year',y='lifeExp',hue='country',style='continent')
```

```python
sns.lineplot(data=df,x='year',y='lifeExp',hue='country',style='continent')
```

```python
# facet plot-> make two graph using row and col and only ->to work relplot
# it will not work with scatterplot and lineplot
sns.relplot(kind='scatter',data=tips,x='total_bill',y='tip',row='day',col='sex')
```

```python
# col wrap
sns.relplot(data=gap,x='lifeExp',y='gdpPercap',kind='scatter',col='year',col_wrap=2)
```

# distribution plot

```python
# figure level-> displot
# axes level-> hisplot->kdeplot->rugplot

```

```python
# plotting univariate
sns.histplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips,x='total_bill',kind='hist')
```

```python
# bins parameter
sns.displot(data=tips,x='total_bill',kind='hist',bins=20)
```

```python
sns.displot(data=tips,x='day',kind='hist',bins=20)
# countplot
```

```python
# hue parameter
sns.displot(data=tips,x='tip',kind='hist',hue='sex')
```

```python
# element -> step
sns.displot(data=tips,x='tip',kind='hist',hue='sex',element='step')
```

```python
titanic=sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic,x='age',kind='hist',element='step',hue='sex')
```

```python
# faceting row and col-> not work on histplot function because they are axis level function
sns.displot(data=titanic,x='age',kind='hist',element='step',col='sex')
```

```python
# kdeplot(kernel density estimation)
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True,height=10,aspect=2)
```

```python
# rugplot-# Rugplot
# Plot marginal distributions by drawing ticks along the x and y axes.
# This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way.
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
```

```python
# bivariate histogram
sns.histplot(data=tips,x='total_bill',y='tip')
```

```python
# bivariate kdeplot-> 2d gaussian
sns.kdeplot(data=tips,x='total_bill',y='tip')
```

# matrix plot

```python
# heatmap->rectangular data
temp=gap.pivot(index='country',columns='year',values='lifeExp')

plt.figure(figsize=(15,15))
sns.heatmap(temp)
```

```python
dfs=gap[gap['continent']=='Europe'].pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=(15,15))
sns.heatmap(dfs,annot=True,linewidths=0.5,cmap='summer')
```

```python
# clustermap-> similar to heatmap -> simlar col together
iris=px.data.iris()
iris
```

```python
sns.clustermap(iris.iloc[:,[0,1,2,3]])
```

# categorial plot

Categorical Scatter Plot->
Stripplot
Swarmplot

Categorical Distribution Plots->
Boxplot
Violinplot

Categorical Estimate Plot -> for central tendency->
Barplot
Pointplot
Countplot

Figure level function -> **catplot**

```python
# strip plot
# axis level function
sns.stripplot(data=tips,x='day',y='total_bill',jitter=False)
```

```python
# catplot
sns.catplot(data=tips,x='day',y='total_bill',kind='strip')
```

```python
# jitter
sns.catplot(data=tips,x='day',y='total_bill',kind='strip',jitter=1.1,hue='sex')
```

```python
# swarmplot
sns.catplot(data=tips,x='day',y='total_bill',kind='swarm',hue='sex')
```

```python
# # categorial distribution plot
# # boxplot
# A boxplot is a standardized way of displaying the distribution
#  of data based on a five number summary (“minimum”, first quartile [Q1],
# median, third quartile [Q3] and “maximum”). It can tell you about your
# outliers and what their values are. Boxplots can also tell you if your
#  data is symmetrical, how tightly your data is grouped and if and how your
# data is skewed.
```

```python
# boxplot
import seaborn as sns

sns.boxplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
# single boxplot-> numerical col
sns.boxplot(data=tips,x='total_bill')
```

```python
# voilion(boxplot+kdeplot)
sns.violinplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
sns.catplot(data=tips,x='day',y='total_bill',hue='sex',kind='violin',split=True)
```

```python
# categorical estimation plot
# barplot
sns.barplot(data=tips,x='day',y='total_bill',hue='sex',errorbar=None,estimator=np.std)
```

```python
# pointplot
sns.pointplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
# countplot
sns.countplot(data=tips,x='sex',hue='day')
```

```python
# faceting with cateplot
sns.catplot(data=tips,x='sex',y='total_bill',col='smoker',kind='bar',row='time')
```

# multigrid

```python
# facegrid
sns.catplot(data=tips,x='sex',y='total_bill',kind='violin',col='day',row='time')
```

```python
g=sns.FacetGrid(data=tips,col='day',row='time',hue='smoker')
g.map(sns.scatterplot,'sex','total_bill')
g.add_legend()
```

```python
# pairplot
sns.pairplot(iris,hue='species')
```

```python
# pairgrid
g=sns.PairGrid(iris,hue='species')

# gmap
g.map(sns.scatterplot)
```

```python
# map_dig and map_offdiag
g=sns.PairGrid(iris,hue='species')
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
g.add_legend()
```

```python
# map_dig and map_upper and maplower
g=sns.PairGrid(iris,hue='species')
g.map_diag(sns.histplot)
g.map_upper(sns.scatterplot)
g.map_lower(sns.boxplot)
g.add_legend()
```

```python
# jointplot
sns.jointplot(data=tips,x='total_bill',y='tip',kind='kde')
```

```python
# joingrid
g=sns.JointGrid(data=tips,x='total_bill',y='tip')
g.plot(sns.scatterplot,sns.histplot)
```
