```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# colored scattered

```python
iris=pd.read_csv('iris.csv')
iris
```

```python
species_numeric, _ = pd.factorize(iris['Species'])
plt.scatter(iris['SepalLengthCm'],iris['PetalWidthCm'],c=species_numeric,cmap='summer',alpha=0.7)
plt.colorbar()
plt.xlabel('sepal length')
plt.ylabel('petalwidth')
plt.show()
```

```python
iris['Species'] = iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
iris.sample(2)
```

# plot size

```python
plt.figure(figsize=(15,7))
plt.scatter(iris['SepalLengthCm'],iris['PetalWidthCm'],c=iris['Species'],cmap='summer',alpha=0.7)
plt.colorbar()
plt.xlabel('sepal length')
plt.ylabel('petalwidth')
plt.show()
```

# annotation

```python
batters=pd.read_csv('batter.csv')
batters.head(5)
```

```python
sample_df=batters.head(100).sample(25,random_state=5)
```

```python
sample_df
```

```python

# annotations
plt.figure(figsize=(15,7))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
plt.show()
```

```python
x=[1,2,3,4]
y=[5,6,7,8]
plt.scatter(x,y)
plt.text(1,5,'point1')
plt.text(2,6,'point2')
plt.text(3,7,'point3')
plt.text(4,8,'point4',fontdict={})
```

# horizontal and vertical line

# 3d surface plot

```python
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
```

```python
xx,yy=np.meshgrid(x,y)
xx,yy.shape
```

```python
z=xx**2+yy**2
z.shape
```

```python
fig=plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
surf = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(surf)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z',color='red')
ax.set_title('surface 3d')
```

```python
z=np.sin(xx) + np.cos(yy)
fig=plt.figure(figsize=(12,8))
fig=plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
surf = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(surf)
```

# countor plot
convert into 2d

```python
fig=plt.figure(figsize=(12,8))
ax=plt.subplot()
surf = ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(surf)
ax.set_xlabel('x')
ax.set_ylabel('y')

ax.set_title('surface 2d')
```

# contour fill

```python
fig=plt.figure(figsize=(12,8))
ax=plt.subplot()
surf = ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(surf)
```

```python
# heatmap is the grid convrt into graph
ipl=pd.read_csv('/content/IPL_Ball_by_Ball_2008_2022.csv')
ipl.head()
```

```python

plt.figure(figsize=(12,8))
plt.imshow(grid)
plt.colorbar()
```
