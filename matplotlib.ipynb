# data visualization on matplotlib

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
plt.style.use('fivethirtyeight')
```

# two types of data
# 1. numeical data
# 2. categorical data

# **2d plot**

```python
# plotting a simple function
price = [40000,20000,22000,340000,500000]
year=[2015,2016,2018,2019,2020]
plt.plot(year,price)
plt.show()
```

```python
# from a pandas dataframe
batsman=pd.read_csv('/content/sharma-kohli.csv')
batsman
```

```python
batsman.head(2)
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])

plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.show()
```

```python
# labels  titles
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.title('rohit sharma vs kholi carrer comparsion')
plt.xlabel('season')
plt.ylabel('runs')
plt.show()
```

```python
# colors(hex) and line(width and style) and marker(size)
plt.plot(batsman['index'],batsman['V Kohli'],color='#F4320B',linestyle='solid',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashdot',linewidth=2)
plt.title('rohit sharma vs kholi carrer comparsion')
plt.xlabel('season')
plt.ylabel('runs')
plt.show()
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='#F4320B',linestyle='solid',linewidth=3,marker='o')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashdot',linewidth=2,marker='o')
plt.title('rohit sharma vs kholi carrer comparsion')
plt.xlabel('season')
plt.ylabel('runs')
plt.show()
```

```python
# length -> location
plt.plot(batsman['index'],batsman['V Kohli'],color='#F4320B',linestyle='solid',linewidth=3,marker='o',label='virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashdot',linewidth=2,marker='o',label='rohit')
plt.legend()
plt.title('rohit sharma vs kholi carrer comparsion')
plt.xlabel('season')
plt.ylabel('runs')
plt.show()
plt.legend(loc='upper right')
```

```python
# limiting area
price = [40000,20000,22000,34000,50000,700000]
year=[2015,2016,2018,2019,2020,2021]
plt.plot(year,price)
plt.xlim(2016,2020)
```

```python
# grid
plt.plot(batsman['index'],batsman['V Kohli'],color='#F4320B',linestyle='solid',linewidth=3,marker='o',label='virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashdot',linewidth=2,marker='o',label='rohit')
plt.legend()
plt.title('rohit sharma vs kholi carrer comparsion')
plt.xlabel('season')
plt.ylabel('runs')
plt.grid()
plt.show()

```

```python
# on panda dataframe
batter=pd.read_csv('/content/batter.csv')
batter.head(2)
```

```python
plt.scatter(batter['avg'],batter['strike_rate'],color='red',marker='D')
plt.title('avg vs strike rate of top 50')
plt.xlabel('avg')
plt.ylabel('strike_rate')
plt.show()
```

```python
# SIZE
tips=sns.load_dataset('tips')
tips
# lower
plt.scatter(tips['total_bill'],tips['tip'],s=tips['size']*20,color='red',marker='o')
plt.title('total bill vs tip')
plt.xlabel('total_bill')
plt.ylabel('tip')
plt.show()
```

```python
# using plt.plot
# faster
plt.plot(tips['total_bill'],tips['tip'],'o')
```

# bar chart

```python
# simple bar chart
children=[10,20,40,10,30]
colors=['red','blue','green','yellow','pink']
plt.bar(colors,children,color='red')
```

```python
# two types of bar chart-> horizontal nad vertical
plt.barh(colors,children,color='red')
```

```python
# colors and label
df=pd.read_csv('/content/batsman_season_record.csv')
df
```

```python
plt.bar(df['batsman'],df['2015'],width=0.5)
```

```python
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.3,color='red')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.3)
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.3)
plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.show()
```

```python
# label overlapping
children=[10,20,40,10,30]
colors=['red red red red red red ','blue blue blueblue blue blue','green greengreen','yellow','pink']
plt.bar(colors,children,color='red')
plt.xticks(rotation='vertical')
```

```python
# stacked bar chart
plt.bar(df['batsman'],df['2015'])
plt.bar(df['batsman'],df['2016'],bottom=df['2015'])
plt.bar(df['batsman'],df['2017'],bottom=df['2016']+df['2015'])
```
